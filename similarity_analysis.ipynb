```python
import pandas as pd
from scipy.stats import pearsonr
```

```python
# load data
similarity_df = pd.read_csv("./Output/posttag_tag_proportions_similarity.csv")
```

```python
# calculate 

pro_data = similarity_df[
    ["pro_phonp", "text_similarity_score"]
].dropna()

pro_r, pro_p = pearsonr(
    pro_data["pro_phonp"],
    pro_data["text_similarity_score"]
)

pro_ci = pearsonr(
    pro_data["pro_phonp"],
    pro_data["text_similarity_score"]
).confidence_interval(confidence_level=0.95)

# anti-ph/onp
anti_data = similarity_df[
    ["anti_phonp", "text_similarity_score"]
].dropna()

anti_r, anti_p = pearsonr(
    anti_data["anti_phonp"],
    anti_data["text_similarity_score"]
)

anti_ci = pearsonr(
    anti_data["anti_phonp"],
    anti_data["text_similarity_score"]
).confidence_interval(confidence_level=0.95)
```

```python
# display results (correlation, p-value, confidence interval)

# Pro-PH/ONP
print("PRO PHONP")
print(f"r = {pro_r:.3f}")
print(f"p = {pro_p:.6g}")
print(f"95% CI = [{pro_ci.low:.3f}, {pro_ci.high:.3f}]")

# Anti-PH/ONP
print("\nANTI PHONP")
print(f"r = {anti_r:.3f}")
print(f"p = {anti_p:.6g}")
print(f"95% CI = [{anti_ci.low:.3f}, {anti_ci.high:.3f}]")
```
