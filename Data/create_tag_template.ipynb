```python
import json
import csv
from pathlib import Path
```

```python
# initialise directories
RAW_DIR = Path("./comment_sections/raw")
TAGGED_DIR = Path("./comment_sections/tagged/pretag")
```

```python
def process_json_file(json_path: Path):
    """
    """
    
    # Load JSON
    with json_path.open("r", encoding="utf-8") as f:
        data = json.load(f)

    # get comments
    comments = data.get("comments", [])

    # Matching output path, preserving subfolder and filename
    relative_path = json_path.relative_to(RAW_DIR)
    output_path = TAGGED_DIR / relative_path.with_suffix(".csv")

    # create output dir if none exists
    output_path.parent.mkdir(parents=True, exist_ok=True)

    # write to csv
    try:
      with output_path.open("w", encoding="utf-8", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["comment", "tag"])

        # write comments to csv
        for comment in comments:
            text = comment.get("text", "")

            writer.writerow([text, ""])

      print(f"Processed: {json_path} -> {output_path}")
        
    # handle exceptions
    except Exception as e:
      print(f'Error processing {json_path.name}: {e}')
```

```python
# Process every JSON file recursively
for json_file in RAW_DIR.rglob("*.json"):
    process_json_file(json_file)
```
